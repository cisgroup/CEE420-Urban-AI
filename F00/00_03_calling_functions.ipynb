# 00-03 - Calling Functions

Almost every line of Python you write this semester will be a **function call**. In 00-02 we used four of them without saying so: `print`, `type`, `int` and `round`. Here we look at what a call actually is, what the things inside the brackets do, and how to find out what an unfamiliar function wants without leaving the notebook.

We also meet a mistake that produces no error at all, which is the most valuable thing in this notebook.

Twenty minutes.

## What a call looks like

A function call is a name, then round brackets, and often something inside the brackets:

```python
round(47.6586, 2)
```

47.66.

The name `round` refers to a piece of code somebody else wrote. The things inside the brackets are the **arguments**, i.e. what we are handing it. And the value that appears is what the function **returns**, i.e. what it hands back.

Those three words are the whole vocabulary: a function takes arguments and returns a value.

Note that the brackets are what makes it happen. They are not punctuation around the name, they are the instruction to run it, and we will see in a moment what happens when you leave them off.

## Arguments, and what happens when you leave one out

Many functions will work with fewer arguments than they accept, because the missing ones have a sensible **default**:

```python
print(round(47.6586, 2))
print(round(47.6586))
```

47.66 and 48.

The second argument said how many decimal places to keep. Leaving it out did not fail; `round` has a default of zero decimal places, i.e. round to a whole number.

Defaults are why you can write `print(x)` without specifying a separator, an end character, or where the output should go, all of which `print` will happily accept if you ever need them.

Note that the order matters when you pass arguments this way. The first is the number, the second is the decimal places, and swapping them asks a completely different question:

```python
print(round(2, 47.6586))
```

That one failed, because the second argument has to be a whole number of decimal places and 47.6586 is not.

But notice how easily it might not have failed. `round(2, 3)` is perfectly legal and returns 2, so a swap between two whole numbers would have given you a wrong answer in silence. Argument order is not a formality.

## Too many arguments

If you give a function more than it knows what to do with, it stops:

```python
round(47.6586, 2, 3)
```

`TypeError: round() takes at most 2 arguments (3 given)`.

That message is unusually helpful, since it tells you both what the function accepts and what you gave it. Most argument mistakes announce themselves this clearly, which is why the silent one in the next section is worth knowing about.

## Naming your arguments

Once a function takes more than two or three arguments, remembering their order becomes unreasonable. So you can name them instead, which is called a **keyword argument**:

```python
print("Princeton", "New Jersey", sep=", ")
```

`Princeton, New Jersey`.

The first two arguments were positional, i.e. identified by where they sit. The third was named, `sep=", "`, which tells `print` what to put between the things it prints, and the default is a single space.

Keyword arguments are how almost every mapping call in this course is written, because a plot has a dozen adjustable things and nobody could remember their order. When you meet `plot(figsize=(8, 8), color="grey", linewidth=0)` in a precept, that is three keyword arguments and nothing more complicated.

The rule: positional arguments come first, keyword arguments after them.

## Functions that belong to a value

There is a second kind of call, written with a dot, and you will use it constantly.

Some functions belong to a particular value, and are called **methods**. You reach them by writing the value, a dot, and the method name:

```python
cafe = "Small World Coffee"

print(len(cafe))
print(cafe.upper())
print(cafe.split())
```

18, then the name in capitals, then `['Small', 'World', 'Coffee']`.

Look at the difference. `len(cafe)` is a plain function taking the string as an argument. `cafe.upper()` is a method belonging to the string, and it needs no arguments at all because it already knows what it is working on.

Note the empty brackets on `cafe.upper()`. They are still required. A method with no arguments is still a call, and the brackets are still what makes it run.

## The mistake that does not fail

Here is the one to remember, because nothing goes red and beginners lose real time to it.

Watch what happens if you forget the brackets:

```python
cafe.upper
```

No error. Instead something like `<built-in method upper of str object at 0x...>`.

Read that carefully. Python did exactly what you asked: you referred to the method without calling it, so it handed you back the method itself, as a value. It is the difference between pointing at a kettle and switching it on.

You will do this to `.head` on a table in the precepts, and you will get a line about a bound method instead of your data, and now you will know why in one second rather than five minutes. **If a cell prints something mentioning "method" or "function" at a hexadecimal address, you left the brackets off.**

## Finding out what a function wants

You do not have to remember any of this, and you should not try. Every function carries its own documentation, and `help` prints it:

```python
help(round)
```

That is the function's own description of itself: what it takes, what is optional, and what it returns.

There is a faster way while you are typing. Put your cursor inside the brackets of a call and press `Shift+Tab`, and the same information appears in a small panel without you having to run anything or leave your place. It is worth building the habit now, because it works on every function in every library you will meet, including the ones nobody has taught you.

Note that some documentation is written for people who already know the library, and can be dense. That is not a failure on your part. Read the first line and the argument list, ignore the rest, and come back if you need to.
