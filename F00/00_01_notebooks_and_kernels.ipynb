## Cells

A notebook is a stack of boxes called **cells**, laid out one after another down the page. There are two kinds.

**Text cells**, like this one, hold writing. They are for explaining things and for your own notes. You can edit one by double-clicking it, and turn it back into neat text by pressing `Shift+Enter`. Try that on this cell if you like; nothing bad happens.

**Code cells** hold Python. They have a slightly different look, usually a grey or dark background, and a small play triangle appears at the left edge when your mouse is near. When you run a code cell, whatever it produces appears directly underneath it.

To run a cell, click it once to select it, then press `Shift+Enter`. That runs the cell and moves you to the next one, which means you can walk down a whole notebook by pressing `Shift+Enter` repeatedly.

Here is a code cell. Run it:

```python
2 + 2
```

Four appeared underneath. That is the whole mechanism, and everything else in this course is a more elaborate version of it.

Note that we did not have to say "print" or "show me". A notebook automatically displays the value of the **last line** of a cell. That is a convenience, and it has one consequence worth knowing now: if a cell has several lines, only the last one shows up on its own.

Let us see that happen:

```python
2 + 2
10 + 10
```

Only 20 appeared. The first line ran, and its answer was thrown away without ever being shown.

When you want to see more than one thing, you have to ask, using `print`:

```python
print(2 + 2)
print(10 + 10)
```

Both, this time. You will forget this at some point and wonder why a cell seems to have skipped something. It did not skip anything; it simply did not show you.

## The number in the brackets

Look to the left of the code cells you have run. Each has a small number in square brackets, `[1]`, `[2]`, `[3]`, counting up in the order you ran them.

That number is not the position of the cell on the page. It is the order in which things actually ran. If a cell has never been run it shows `[ ]`, empty. If it is running right now it shows `[*]`, a star, which on a slow cell you will see for a second or two.

This distinction between where a cell sits and when it ran is the entire subject of this notebook, and we are about to make it hurt.

## Names, and the thing that remembers them

Python lets you give a name to a value, which we call **assignment**. The name goes on the left, the value on the right, and the equals sign points right to left, i.e. it means "put that in here" rather than "these are equal".

Let us name something:

```python
answer = 2 + 2
```

Nothing appeared, and that is correct. We did not ask for the value, we stored it. The name `answer` now refers to 4.

Note that this is the second reason a cell can look like it did nothing. Sometimes it printed nothing because you stored the result, and sometimes because there was nothing to show. Neither is a failure.

Now the name exists, and any later cell can use it:

```python
answer * 10
```

Forty, as you would expect.

Now stop and ask a question that sounds trivial and is not. **Where is that 4 being kept?**

Not in the notebook. The notebook is a document, i.e. paper. Behind it sits a separate program called the **kernel**, which is the Python that actually runs your code, and the kernel is what remembers that `answer` is 4. When you chose an environment at the top of this window, you were choosing which kernel to start.

The kernel remembers every name you have given it, for as long as it is running, and it does not care what is written on the page. It only knows what it has been told to run.

## The experiment

This is the part to do rather than read. It takes two minutes, and it is the single most useful thing in this notebook.

**Step one.** Scroll up to the cell that reads `answer = 2 + 2`. Change the 2 + 2 to `5`, so the line reads `answer = 5`. Do not run it. Just change the text.

**Step two.** Come back down to the cell that reads `answer * 10` and run it again with `Shift+Enter`.

**What do you predict?** The page above now says `answer = 5`, so you might reasonably expect 50.

Run it and see.

It still says 40.

Nothing is broken. You changed the paper, and the kernel does not read the paper. It is still holding the 4 you gave it earlier, and it will keep holding it until you run that cell and tell it otherwise.

**Step three.** Now go back up and run the `answer = 5` cell, then come back down and run `answer * 10` again. Now it says 50, because now the kernel has been told.

Look at the numbers in the brackets while you do this. They keep counting up, and they no longer match the order of the cells on the page. That mismatch is the honest record of what happened, and it is why the brackets exist.

## Why this matters more than it sounds

Everybody meets this eventually, usually like this. You write some code, it works. You tidy it up, rename a few things, run only the cell you are working on, and everything is fine. You close the laptop.

The next day you open the notebook, run it from the top, and it collapses, because it never actually worked in the order it is written. It only worked in the order you happened to click.

That is why the fix is a habit rather than a technique. Whenever a notebook confuses you, prove it is honest by starting again from nothing. The command lives in the Run menu and is called **Restart Kernel and Run All Cells**. It throws away everything the kernel remembers and runs the notebook from the top, in the order it is written, which is the only order anybody else will ever run it in.

Do that now with this notebook. Everything should come back exactly as it was, and every number in brackets should count neatly from one.

## When a name does not exist

One more thing to see on purpose, because you will see it by accident.

If you ask the kernel for a name it has never been given, it stops and tells you so. That is not a crash and nothing is damaged; it is Python saying it does not know what you mean.

The next cell asks for a name that has never been defined anywhere. It will produce a red block of text, and that is the intended outcome:

```python
prnceton_boundary
```

Red text, ending in a line that reads `NameError: name 'prnceton_boundary' is not defined`.

Read that last line and nothing else for now. It names the kind of problem, `NameError`, and it tells you exactly which name it could not find. In this case the reason is a typo, i.e. a missing `i` in Princeton, and the fix is one keystroke.

There are two quite different reasons you will see a `NameError` this semester, and it is worth being able to tell them apart:

- The name is misspelled, as here.
- The name is spelled correctly, but the cell that creates it has not been run yet. This happens constantly when you open a notebook fresh and start clicking in the middle of it.

For the second one, the fix is the habit from the previous section: restart and run all, and everything the notebook needs gets created in order.
