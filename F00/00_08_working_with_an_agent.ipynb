This course expects you to use AI coding agents, and expects you to be good at it. Being good at it is not about knowing which tool to buy. It is about two habits, both of which are cheap and neither of which is obvious.

In this notebook we ask an agent for the same thing twice, badly and then well, and we look at what changes. We use a question you can already answer yourself from 00-05, because the only way to learn to check an answer is to check one you already know.

Twenty minutes, and no agent is required to read it.

````python
import pandas as pd

food = pd.read_csv("data/food_table.csv")
print(f"{len(food)} places to eat")
````

## What an agent is, and what it is not

A coding agent is a program that writes code from a description. Modern ones can read the files in your project, write into your notebook, and run what they wrote.

It is worth being precise about what that does and does not mean. An agent is very good at the mechanical parts: remembering the exact name of a method, writing the loop, getting the brackets right. It has no idea what your data means, whether the question you asked is the question you wanted answered, or whether the number it produced is plausible for the town you are studying.

So it removes the part of programming that was never the hard part, and leaves the hard part exactly where it was. That is why this course grades you on what you checked rather than on what you produced.

## The vague request

Suppose you want to know how many cafes there are, and you ask for it the way you would ask a person: "count the cafes".

Here is code of the kind that request produces:

````python
count = len(food[food["name"].str.contains("Coffee|Cafe|Café", case=False, na=False)])
print(f"cafes found: {count}")
````

Seven.

That code is not wrong, exactly. It searched the names for coffee-related words, which is a reasonable reading of "count the cafes" if you did not know the data had an `amenity` column. It runs, it produces a plausible number, and nothing about it looks suspicious.

But we know from 00-05 what the answer is:

````python
print(food["amenity"].value_counts())
````

Fifteen cafes.

So the vague request produced a confident, plausible answer that is wrong in both directions at once, and for reasons nobody would guess by reading the code.

It **missed** eight real cafes, including both branches of Starbucks, `Ellinikon`, `Kung Fu Tea` and `Rojo's Roastery`, because none of them has a coffee word in its name. And it **wrongly included** `Chuck's Spring Street Cafe`, which has "Cafe" in its name and is tagged in the data as something else entirely.

Note the shape of this failure, because it is the shape of most of them. Nothing broke. The number was in the right range. Only somebody who already knew the answer would have caught it, and the specific way it failed, i.e. that a cafe called Starbucks does not contain the word cafe, is obvious in hindsight and invisible in advance.

## The specification

The fix is to say what you want precisely enough that a wrong answer would be obviously wrong. Three lines are enough:

```text
PLACE:  which data, which file, which column
TASK:   what to compute, in what units
OUTPUT: what you want back
```

Filled in for this question:

```text
PLACE:  the DataFrame `food`, loaded from data/food_table.csv
TASK:   count the rows where the column `amenity` is exactly "cafe"
OUTPUT: the count as a single integer, printed
```

That request cannot reasonably produce the name-searching version, because it names the column and says "exactly". Here is what it gives instead:

````python
count = int((food["amenity"] == "cafe").sum())
print(f"cafes: {count}")
````

Fifteen, agreeing with `value_counts`.

The important thing is not that this version is shorter. It is that **writing the specification is what made you decide what you actually meant.** You had to know there was an `amenity` column, and that it should match exactly rather than approximately. Having thought that through, you would have caught the first version immediately.

That is why the specification is for you rather than for the agent. If you cannot say what a correct answer looks like, you cannot recognise one when it arrives.

## Why units belong in a specification

One more thing to put in a specification, and it is the one that bites hardest in this course.

A number without a unit is not an answer. If you ask an agent for "everything within 400 of this point", it will produce code that runs, and whether 400 meant metres, feet or degrees of longitude depends entirely on what coordinate system the data happened to be in. In the first precept you will see this produce answers of 92, 51 and 55 for the same question, all from code that ran without complaint.

So a specification that says "within a 400 metre straight line, computed in a coordinate system measured in metres" is not pedantry. It is the difference between an answer and a number.

## The audit

The specification is half the habit. The other half is what you do when the answer arrives.

**Read the code before you run it.** Not to verify every character, but to see whether it did roughly what you asked. In the vague example above, one glance at `str.contains` tells you it is searching names, which is not what the `amenity` column is for.

**Check the answer against something you already know.** A total, a count from another method, a rough figure from memory, anything independent. `value_counts` gave us 15 without any agent involved, which is why we could catch the 12.

**Ask whether the number is plausible.** Princeton is a town you can walk across. If something reports 513 square kilometres, or 92 restaurants inside a 400 metre circle, no code review is needed; the number is telling you.

And when your answer differs from somebody else's, the question is not who is right. It is whether you can each say why. A difference you can account for is a finding. A difference you cannot account for is a bug that has not been found yet.

## The AI note

Every piece of work you hand in this semester carries three lines saying what you used, what it did, and what you checked:

```text
AI note
Tools:    Antigravity with Gemini, for the filtering code
They did: wrote the value_counts call and the filter
Verified: the count agrees with value_counts, and the six categories
          add to 92, which is the number of rows
```

That is not a confession and it is not a formality. Written honestly, it is a record of your own reasoning, and it makes the work yours in the sense the honour pledge means. Written dishonestly, it is the one thing in this course that is actually an integrity problem.
