# 00-04 - Lists and Loops

So far every name has held one thing. Real data comes in quantities, so we need a way to hold many things at once and a way to do something to each of them in turn.

That is a **list** and a **loop**, and between them they are most of the programming you will ever do. We build both out of the six kinds of place you can eat in Princeton.

## A list

A list is written with square brackets and commas between the items:

```python
amenities = ["restaurant", "fast_food", "cafe", "ice_cream", "pub", "bar"]
counts = [52, 16, 15, 6, 2, 1]

print(amenities)
print(f"{len(amenities)} kinds, {sum(counts)} places in total")
```

Six kinds, 92 places, which is the number of places to eat in our Princeton data.

Two functions did the work there. `len` counts the items, which we met on a string in 00-03 and which works the same way here. And `sum` adds up a list of numbers, which only makes sense on numbers, so `sum(amenities)` would fail.

Note that a list keeps its order. These two lists line up: `counts[0]` is the count of `amenities[0]`, and that correspondence is entirely our responsibility. Nothing in Python ties them together, and if we sorted one and not the other the data would be quietly ruined. In 00-05 we meet a structure that does not have this problem.

## Getting one item out

You reach an item by its position in square brackets, and **positions start at zero**:

```python
print(amenities[0])
print(amenities[1])
```

`restaurant` and `fast_food`.

Counting from zero feels wrong for about a week and then becomes invisible. The useful way to think about it is that the number is not "which item" but "how many to step past", so the first item is zero steps in.

The consequence is that the last item of a six-item list is at position five, not six. Asking for six is an error:

```python
amenities[6]
```

`IndexError: list index out of range`.

**`IndexError` means you asked for a position that does not exist.** With six items the valid positions are 0 to 5, and this is such a common mistake that it has a name, the off-by-one error.

Python offers a shortcut so you rarely have to count. A negative position counts backwards from the end:

```python
print(amenities[-1])
print(amenities[-2])
```

`bar` and `pub`, i.e. the last and the second to last. `[-1]` is the idiom for "the last one" and you will see it constantly.

## Several items at once

Two positions separated by a colon give you a **slice**, i.e. a smaller list:

```python
print(amenities[0:3])
print(amenities[3:])
```

The first three, then everything from position three onwards.

Note the asymmetry: `[0:3]` gave positions 0, 1 and 2, and stopped before 3. The start is included and the end is not. That looks arbitrary and has a convenient consequence, namely that `[0:3]` and `[3:]` between them cover the whole list exactly once, with no overlap and nothing missed.

Leaving out the end means "to the end", and leaving out the start means "from the beginning".

## Doing something to each item

Printing items one at a time by position would be tedious and would break the moment the list changed length. A **loop** goes through every item, whatever the length:

```python
for kind in amenities:
    print(kind)
```

Six lines, one per item.

Read the first line as an instruction: for each item in `amenities`, give it the temporary name `kind`, and then do what follows. The name `kind` is ours to choose; it did not exist before and it takes each value in turn.

**The indentation is not decoration.** The spaces at the start of the second line are how Python knows that line belongs to the loop. Other languages use brackets for this; Python uses the layout, which means the layout is part of the meaning.

Miss it out and Python stops immediately:

```python
for kind in amenities:
print(kind)
```

`IndentationError: expected an indented block`.

The fix is always the same: indent the lines that belong to the loop, consistently, by four spaces. Your editor will do it for you when you press Tab, and it will usually indent automatically after a line ending in a colon.

## Looping over two lists together

We have kinds in one list and counts in another, and we want them side by side. `zip` walks two lists in step:

```python
for kind, count in zip(amenities, counts):
    print(f"{kind}: {count} places")
```

Six lines pairing each kind with its count, which is the first output in this notebook you might actually put in front of somebody.

Note that the loop line now names two things, `kind` and `count`, because `zip` hands back a pair each time round.

## Making a decision inside a loop

A loop that treats every item identically is useful. A loop that decides is more useful. `if` runs a block only when something is true:

```python
for kind, count in zip(amenities, counts):
    if count < 5:
        print(f"{kind}: only {count}, too few to say anything about")
    else:
        print(f"{kind}: {count}")
```

The three rare kinds are flagged and the common ones are printed plainly.

Note the two levels of indentation. The `if` belongs to the loop, so it is indented once; the `print` belongs to the `if`, so it is indented twice. The layout is the structure.

Note also what we just did, which is more interesting than the syntax. We drew a line at five and declared everything below it too sparse to interpret. That threshold is a judgement, not a fact, and the ice cream shops did not become unimportant because we chose five. Every analysis in this course contains decisions like that one, and the honest thing is to say where they are.

## Building up an answer

The last pattern, and the one that appears most in real work: start with an empty list, and add to it as you go.

```python
common = []

for kind, count in zip(amenities, counts):
    if count >= 10:
        common.append(kind)

print(common)
print(f"{len(common)} of {len(amenities)} kinds have ten or more places")
```

Three kinds out of six.

`append` is a method belonging to the list, as in 00-03, and it adds one item to the end. The empty list `[]` before the loop is the container we fill.

This shape, i.e. empty container, loop, condition, append, is worth recognising because you will write it many times. In 00-05 we meet a tool that does the same job in one line, but knowing the long way first means the short way is a convenience rather than a magic word.
